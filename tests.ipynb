```python
from main import DNASequence, RNASequence, AminoAcidSequence
from main import filter_fastq
```

```python
# DNASequence
dna = DNASequence('ATGCGT')

print(dna)                    # Sequence: ATGCGT
print(len(dna))               # 6
print(dna[0])                 # A
print(dna[1:4])               # TGC
print(dna.is_valid())         # True
print(dna.reverse())          # TGCGTA
print(dna.complement())       # TACGCA
print(dna.reverse_complement())  # ACGCAT
print(dna.transcribe())       # AUGCGU

print(DNASequence('AUGC').is_valid())  # False — U не допустим в ДНК
```

```python
# RNASequence
rna = RNASequence('AUGCGU')

print(rna)                    # Sequence: AUGCGU
print(len(rna))               # 6
print(rna.is_valid())         # True
print(rna.complement())       # UACGCA
print(rna.reverse_complement())  # ACGCAU

print(RNASequence('ATGC').is_valid())  # False — T не допустим в РНК
```

```python
# AminoAcidSequence
prot = AminoAcidSequence('MKTLL')

print(prot)                   # Sequence: MKTLL
print(len(prot))              # 5
print(prot[0])                # M
print(prot.is_valid())        # True
print(prot.triple_alphabet()) # MetLysThrLeuLeu

print(AminoAcidSequence('MKTX').is_valid())  # False — X не допустим
```
